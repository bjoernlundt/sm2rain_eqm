```python
import xarray as xr
import os
```

```python
os.chdir('c:\\Users\\bjoer\\Documents\\HiWi\\data_sm2rain_daily\\gpm_sm2rain')
```

```python
# List all .nc files in the current directory
nc_files = [file for file in os.listdir() if file.endswith('.nc')]

# Load and concatenate all .nc files along the time dimension
datasets = [xr.open_dataset(file) for file in nc_files[1:]]
combined_dataset = xr.concat(datasets, dim='time')
```

```python
combined_dataset = combined_dataset.assign_coords(
    lon=combined_dataset["Longitude"].isel(time=0),  # Take first time step to get lon
    lat=combined_dataset["Latitude"].isel(time=0)   # Take first time step to get lat
)
```

```python
combined_dataset = combined_dataset.drop_vars(["Longitude", "Latitude"])
```

```python
combined_dataset = combined_dataset.set_coords(["lon", "lat"])
```

```python
combined_dataset_subset = combined_dataset.where(
    (combined_dataset.lon >= -3) & (combined_dataset.lon <= 1) & 
    (combined_dataset.lat >= 11) & (combined_dataset.lat <= 14), 
    drop=True
)
```
